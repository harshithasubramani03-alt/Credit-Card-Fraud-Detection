```python
# ============================================================
# Credit Card Fraud Detection
# Step 1: Exploratory Data Analysis (EDA)
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

import warnings
warnings.filterwarnings('ignore')

print("Libraries Imported Successfully ✅")
```

```python
# Load the dataset
df = pd.read_csv(r'C:\Users\harsh\Downloads\fraude detection\creditcard.csv')

print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
df.head()
```

```python
print("=== Dataset Info ===")
df.info()
```

```python
print("=== Statistical Summary ===")
df.describe()
```

```python
print("=== Missing Values ===")
print(df.isnull().sum())

# Visualize
plt.figure(figsize=(14,4))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False, cmap='viridis')
plt.title('Missing Data Heatmap')
plt.show()
```

```python
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
print("After removing duplicates:", df.shape)
```

```python
print("=== Class Distribution ===")
print(df['Class'].value_counts())
print("\nFraud %:", round(df['Class'].mean() * 100, 4), "%")

plt.figure(figsize=(6,4))
sns.countplot(x='Class', data=df, palette='Set2')
plt.title('Fraud vs Legitimate Transactions')
plt.xticks([0,1], ['Legitimate (0)', 'Fraud (1)'])
plt.ylabel('Count')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14,5))

# Fraud vs Legit Amount
sns.boxplot(x='Class', y='Amount', data=df, ax=axes[0], palette='Set2')
axes[0].set_title('Transaction Amount by Class')
axes[0].set_xticklabels(['Legitimate', 'Fraud'])

# Amount distribution
sns.histplot(df[df['Class']==0]['Amount'], bins=50, ax=axes[1], color='green', label='Legit', alpha=0.6)
sns.histplot(df[df['Class']==1]['Amount'], bins=50, ax=axes[1], color='red', label='Fraud', alpha=0.6)
axes[1].set_title('Amount Distribution')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14,5))

sns.histplot(df[df['Class']==0]['Time'], bins=50, ax=axes[0], color='green', label='Legit', alpha=0.6)
sns.histplot(df[df['Class']==1]['Time'], bins=50, ax=axes[0], color='red', label='Fraud', alpha=0.6)
axes[0].set_title('Time Distribution by Class')
axes[0].legend()

sns.boxplot(x='Class', y='Time', data=df, ax=axes[1], palette='Set2')
axes[1].set_title('Time by Class')
axes[1].set_xticklabels(['Legitimate', 'Fraud'])

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20,10))
sns.heatmap(df.corr(), cmap='coolwarm', annot=False, linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```
