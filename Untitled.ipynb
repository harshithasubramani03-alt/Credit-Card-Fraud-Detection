```python
import pandas as pd
credit_card_data = pd.read_csv(r'C:\Users\harsh\Downloads\fraude detection\creditcard.csv')
```

```python
legit = credit_card_data[credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class == 1]
```

```python
legit_sample = legit.sample(n=492)
```

```python
new_dataset = pd.concat([legit_sample, fraud], axis=0)
```

```python
new_dataset.head()
```

```python
new_dataset.tail()
```

```python
new_dataset['Class'].value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

```python
# Option 1: use columns= (recommended)
X = new_dataset.drop(columns='Class')
Y = new_dataset['Class']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42, stratify=Y)

print(X_train.shape, X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, Y_train)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, Y_train)
```

```python
from sklearn.metrics import accuracy_score

Y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(Y_test, Y_pred))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(Y_test, Y_pred))
print(confusion_matrix(Y_test, Y_pred))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Fraud vs Non-Fraud distribution
sns.countplot(x='Class', data=credit_card_data)
plt.title('Fraud vs Legitimate Transactions')
plt.show()

# Correlation heatmap
plt.figure(figsize=(20,10))
sns.heatmap(credit_card_data.corr(), cmap='coolwarm')
plt.show()
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report

model_xgb = XGBClassifier(n_estimators=100, random_state=42)
model_xgb.fit(X_train, Y_train)

Y_pred_xgb = model_xgb.predict(X_test)
print(classification_report(Y_test, Y_pred_xgb))
```

```python
import pickle

with open('fraud_model.pkl', 'wb') as f:
    pickle.dump(model_xgb, f)

print("Model saved!")
```

```python
# Load and test
with open(r'C:\Users\harsh\Downloads\fraude detection\fraud_model.pkl', 'rb') as f:
    loaded_model = pickle.load(f)

print("Model loaded successfully! ✅")
print("Test accuracy:", loaded_model.score(X_test, Y_test))
```
